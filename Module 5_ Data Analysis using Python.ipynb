```python
df['age'] = df['age'].fillna(df['age'].median())

df['likes'] = df['likes'].fillna(df['likes'].median())

df['comments'] = df['comments'].fillna(df['comments'].median())

df['shares'] = df['shares'].fillna(df['shares'].median())
```

```python
df['gender'] = df['gender'].fillna(df['gender'].mode()[0])
df['sentiment'] = df['sentiment'].fillna(df['sentiment'].mode()[0])
```

```python
print("Duplicates:", df.duplicated().sum())
df = df.drop_duplicates()
print("Duplicates after cleaning:", df.duplicated().sum())
```

```python
df['posted_at'] = pd.to_datetime(df['posted_at'])
df['gender'] = df['gender'].str.strip().str.title()
```

```python
print("Negative Likes:", (df['likes'] < 0).sum())
print("Negative Comments:", (df['comments'] < 0).sum())
print("Negative Shares:", (df['shares'] < 0).sum())
df.loc[df['likes'] < 0, 'likes'] = np.nan
df.loc[df['comments'] < 0, 'comments'] = np.nan
df.loc[df['shares'] < 0, 'shares'] = np.nan
df['likes'] = df['likes'].fillna(df['likes'].median())
df['comments'] = df['comments'].fillna(df['comments'].median())
df['shares'] = df['shares'].fillna(df['shares'].median())
```

```python
df['hashtag_count'] = df['hashtags'].str.split().str.len()
df['sentiment'] = df['sentiment'].str.strip().str.lower()
df.head()
```

```python
df['engagement_score'] = (
    df['likes'] +
    df['comments'] +
    df['shares']
)
```

```python
df['hashtag_count'] = (
    df['hashtags']
    .fillna('')
    .str.split()
    .str.len()
)
```

```python
df['log_likes'] = np.log1p(df['likes'])
df['log_comments'] = np.log1p(df['comments'])
df['log_shares'] = np.log1p(df['shares'])
df['log_impressions'] = np.log1p(df['impression_count'])
```

```python
post_type_summary = df.groupby('post_type').agg({
    'likes': 'mean',
    'comments': 'mean',
    'shares': 'mean',
    'engagement_score': 'mean'
})
print("Post Type Summary:")
print(post_type_summary)
```

```python
country_summary = df.groupby('country').agg({
    'impression_count': 'sum',
    'likes': 'mean',
    'comments': 'mean',
    'shares': 'mean',
    'engagement_score': 'mean'
})

print("Country Summary:")
print(country_summary)
```

```python
sentiment_summary = df.groupby('sentiment').agg({
    'likes': 'mean',
    'comments': 'mean',
    'shares': 'mean',
    'engagement_score': 'mean'
})

print("Sentiment Summary:")
print(sentiment_summary)
```

```python
stats_columns = [
    'likes',
    'comments',
    'shares',
    'watch_time_sec',
    'engagement_rate',
    'follower_count'
]

# Mean
print("Mean:")
print(df[stats_columns].mean())

# Median
print("\nMedian:")
print(df[stats_columns].median())

# Mode
print("\nMode:")
print(df[stats_columns].mode().iloc[0])

# Standard Deviation
print("\nStandard Deviation:")
print(df[stats_columns].std())

# Variance
print("\nVariance:")
print(df[stats_columns].var())

# Percentiles
print("\nPercentiles:")
print(df[stats_columns].quantile([0.25, 0.50, 0.75, 0.90]))

# Skewness
print("\nSkewness:")
print(df[stats_columns].skew())

# Kurtosis
print("\nKurtosis:")
print(df[stats_columns].kurt())
```

Behavioral Insights

Best time of day for impressions

There is an important limitation in this dataset: posted_at contains dates only, without a time component. Therefore, all records appear at 00:00 after conversion.

So you cannot reliably determine the best time of day for impressions from this dataset.

Sentiment Analysis

Which sentiment performs best?

Insight: Surprisingly, negative sentiment posts have the highest average engagement rate in this dataset.

Behavior of negative and neutral sentiment posts

Negative posts: Average engagement rate ≈ 1.04%, with approximately 10,211 average likes.
Neutral posts: Average engagement rate ≈ 0.99%, with approximately 9,918 average likes.
Negative posts therefore show slightly higher engagement than neutral and positive posts.
Neutral posts have a slightly higher average number of comments than negative posts.
Overall Conclusion

The analysis shows that video content generates the highest engagement among post types, while food is the strongest-performing content category. Brazil has the highest average engagement rate among the countries analyzed. Engagement varies across age groups, with users aged 45–54 showing the highest average engagement. Verified accounts show slightly higher engagement than non-verified accounts. Mobile users have the highest average watch time. Negative-sentiment posts show the highest average engagement rate, followed by neutral and positive posts. The dataset does not contain actual posting times, so a time-of-day impression analysis cannot be performed reliably.
